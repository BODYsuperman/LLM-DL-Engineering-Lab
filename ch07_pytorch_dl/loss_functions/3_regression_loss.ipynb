```python
import torch
import torch.nn as nn
from requests.packages import target
```

```python
input = torch.randn(3, 5)

target = torch.randn(3, 5)

print(input)
print(target)
```

```python
#MSE L2 Loss

mse_loss = nn.MSELoss()
print(mse_loss(input, target))

#MAE L1 Loss

mae_loss = nn.L1Loss()
print(mae_loss(input, target))

# Smooth L1 Loss

sm_l1_loss = nn.SmoothL1Loss()
print(sm_l1_loss(input, target))
```
