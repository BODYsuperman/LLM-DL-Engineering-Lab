```python
import torch
import torch.nn as nn
```

```python
rnn = nn.RNN(input_size=8, hidden_size= 16, num_layers=2 )
```

```python
#Input: (L, N, H_in) when batch_first=False (default)
#Input: (N, L, H_in) when batch_first=True
#Input[seq_len, batch_size, input_size]
#hx:init hidden layer [num_layers, batch_size, hidden_size]
#output:[seq_len, batch_size, hidden_size]
#hn: hidden layer[num_layers, batch_size, hidden_size]
input = torch.rand(1, 3, 8)

hx = torch.rand(2, 3, 16)

```

```python
output, hn = rnn(input, hx)
print(output.shape)
print(hn.shape)
```
