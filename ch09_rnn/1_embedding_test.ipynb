```python
import torch
import torch.nn as nn
```

```python
import jieba
```

```python
text = "一个词的意义不是由一个符号或一个维度单独表示，而是由向量的许多维度共同激活、共同组合来表示；同时每个维度又参与许多不同词的表示。语义信息分布在整个向量的模式中，而不是集中在某个局部单元里。"

torch.manual_seed(42)
stop_words = {"的","是","而","由",",","。","，","、"}
```

```python
words = [word for word in jieba.lcut(text) if word not in stop_words]

vocab = list(set(words))

word2idx = dict()
print(vocab)
```

```python
for idx, word in enumerate(vocab):
    word2idx[word] = idx

print(word2idx)
emded = nn.Embedding(num_embeddings=len(vocab), embedding_dim=5)


#forward propagation according to idx we can get corresponding vector
for idx, word in enumerate(vocab):
    word_vec = emded(torch.tensor(idx))
    print(f"idx:{idx:>2}:{word:8}\t {word_vec.detach().numpy()}")

```
