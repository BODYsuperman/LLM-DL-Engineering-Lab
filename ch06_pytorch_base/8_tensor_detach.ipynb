```python
import torch
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x.detach()

print(x)
print(y)

print(x.requires_grad)
print(y.requires_grad)
```

```python
print(x.untyped_storage().data_ptr())
print(y.untyped_storage().data_ptr())
```

```python
print(x)
y.zero_()
print(x)
```

```python
z1 = x**2
z2 = y**2

print(z1)
print(z2)

print(z1.requires_grad)
print(z2.requires_grad)
```

```python
print(z1.sum().backward().grad)
print(z2.sum().backward().grad)
print(z2.sum().backward())
```

```python
x = torch.ones(2, 2, requires_grad=True)
y = x * x
print(x)
print(y)
print(x.requires_grad)
print(y.requires_grad)
```

```python
u = y.detach()
print(u)
print(x.requires_grad)
```

```python
z = u * x
print(z)

print(z.sum().backward().grad)
```

```python
print(x.grad == u.grad)
print(x.grad)
```
