```python
import torch
import numpy as np
```

### Tensor type conversion

```python
tensor1 = torch.randint(0, 2, (5, 3))
print(tensor1)
print(tensor1.dtype)

```

```python
tensor2 = tensor1.type(torch.float32)
print(tensor2)
print(tensor2.dtype)
```

```python
tensor1 = tensor1.to(torch.complex64)
print(tensor1)
print(tensor1.dtype)
```

### Conversion between tensor and ndarray

```python
#uniform distribution over (0, 1)
np.set_printoptions(precision=4)
#1. tensor -> ndarray
tensor1 = torch.rand(3,2)
print(tensor1)
print(type(tensor1))

numpy_array = tensor1.numpy()
print(numpy_array)
print(type(numpy_array))

print()

tensor1[:, 0] = 10
print(tensor1)
print(numpy_array)
```

```python
#use copy() to avoid shared memory

numpy_array2 = tensor1.numpy().copy()
tensor1[:, 0] = -1

print(tensor1)
print(numpy_array)
print(numpy_array2)
```

```python
#2. ndarray-> tensor

ndarray1 = np.random.randn(5,3)
print(ndarray1)
tensor1 = torch.from_numpy(ndarray1)
print(tensor1)
```

```python
ndarray1[:, 0] = 5
print(ndarray1)
print(tensor1)

tensor2 = torch.from_numpy(ndarray1.copy())

ndarray1[:, 0] = 7
print(ndarray1)
print(tensor1)
print(tensor2)
```

```python
#use tensor -> Tensor
ndarray1 = np.random.randn(5,3)
print(ndarray1)
tensor1 = torch.tensor(ndarray1)
print(tensor1)
```

```python
ndarray1[:, 0] = 5
print(ndarray1)
print(tensor1)
```

### tensor convert to scalar


```python
tensor1 = torch.tensor(1)
tensor2 = torch.tensor([[2]])
print(tensor1)
print(tensor1.item())

print(tensor2)
print(tensor2.item())
```
