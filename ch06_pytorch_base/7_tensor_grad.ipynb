```python
import torch
```

```python
# define input
x = torch.rand((2,2))
y = torch.rand((2,2))
print(x.grad)
print(y.grad)

#initialize the weight matrix w and b

w = torch.rand(2,2, requires_grad=True)
b = torch.rand(2,2, requires_grad=True)

z = w * x + b

print(z)
loss = torch.nn.MSELoss()

loss_v = loss(z, y)
print(loss_v)
#back propagation
loss_v.backward()

print("the grad of w is ", w.grad)
print("the grad of b is ", b.grad)
```

```python
#x , w, b , y are the most basic nodes are not calculated from other nodes

print(w.is_leaf)
print(b.is_leaf)
print(z.is_leaf)
print(y.is_leaf)
print(loss_v.is_leaf)
```
