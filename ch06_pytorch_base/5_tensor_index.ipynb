```python
import  torch
```

```python
tensor1 = torch.randint(1, 10, (3, 5, 4))
print(tensor1)

```

```python
#simple index

print(tensor1[2,1 , 3])
print(tensor1[1])

print(tensor1[2, 3])
```

```python
print(tensor1[:, 3])
```

```python
print(tensor1[1:])

print(tensor1[-1, 1:4, 0:3])
```

```python
print(tensor1[-1, 1:4:2, 0:3])
```

```python
#list index

#first param is matrix and 2nd is the line in correspond matrix

print(tensor1[[0,1],[3,2]])

print(tensor1[[0,1],[3,2], [1,1 ]])

print(tensor1[[[0],[1]], [1,2]])
```

```python
mask = tensor1[:, :, 0] > 5

print(mask)

print(tensor1[mask])
```

```python
mask = tensor1[:, 1, :] > 5

print(mask)

print(tensor1[mask])
```
