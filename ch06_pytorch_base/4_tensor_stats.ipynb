```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 2, 4)).float()
print(tensor1)
```

```python
#sum
#dim means the dimension of the tensor
# 0 is the 1st , 1 is the 2nd, 2 is the 3rd
#what we set the correspond dimension will be combined

print(sum(tensor1))
print()

print(tensor1.sum(dim=1))

print(tensor1.sum(dim=2))
```

```python
#mean
print(tensor1.mean())
```

```python
print(tensor1.mean(dim=2))
```

```python
#std
print(tensor1.std())
```

```python
#min max

print(tensor1.min())

print(tensor1.min(dim=2))

print(tensor1.argmin())

print(tensor1.argmin(dim=2))

print(tensor1.argmax(dim=2))
print(tensor1.max())
```

```python
# unique remove duplicate

print(tensor1)
print(tensor1.unique())

```

```python
#sort

print(tensor1.sort())
print(tensor1.sort(dim=1))
```
