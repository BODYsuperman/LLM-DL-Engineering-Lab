## 1.The four basic arithmetic operations

```python
import torch
import numpy as np
```

```python
tensor1 = torch.randint(0, 10, [2,3])
print(tensor1)
```

```python
print(tensor1 + 10)

print(tensor1)
```

```python
print(tensor1.add(10))
print(tensor1)
```

```python
print(tensor1.add_(10))
print(tensor1)
```

```python
print(tensor1.sub_(10))
```

```python
# power function

print(tensor1.pow_(2))

print(tensor1.sqrt())

print(tensor1**2)
```

```python
tensor1 = torch.tensor([1.0, 2.0, 3.0])
print(tensor1)
print(2.718 ** tensor1)

print(tensor1.exp())

print(tensor1)
```

```python
tensor1 = torch.tensor([1.0, 2.0, 3.0])
print(tensor1)
print(2.718 ** tensor1)

print(tensor1.exp_())

print(tensor1)
```

```python
## Hadamard product or Element-wise multiplication
tensor1 = torch.randint(0, 5, [2,2])
tensor2 = torch.randint(0, 5, [2,2])
print(tensor1)
print(tensor2)

tensor3 = tensor1 * tensor2
print(tensor3)
print(tensor1.mul(tensor2))

```

```python
#(B, m, n) @ (B, n, p) → (B, m, p)
A = torch.randn(4, 3, 4)   # (B=4, m=3, n=5)
B = torch.randn(4, 4, 2)   # (B=4, n=5, p=2)

C = torch.matmul(A, B)     # (4, 3, 2)
print(A)
print(B)
print()

print(C)
```

```python
A = torch.randn(4, 3, 5)   # batch=4
B = torch.randn(1, 5, 2)   # batch=1

#since batch A is different with batch B and batch B is 1, will lead to broadcast
#batch 0: A[0] @ B[0]
# batch 1: A[1] @ B[0]   ← 还是 B[0]，同一份数据
# batch 2: A[2] @ B[0]   ← 还是 B[0]
# batch 3: A[3] @ B[0]   ← 还是 B[0]
```

```python
#randn is equivalent to randint(0,1 , (size)) uniform distribution over(0,1)
A = torch.randn(2, 3)
B = torch.randn(3,2)
print(A @ B)

print(A.matmul(B))

print(A.mm(B))

```

```python
# save the memory via reusing variable without allocating new memory space

X = torch.randint(1, 10, (3, 2, 4))
Y = torch.randint(1, 10, (1, 4, 5))

print(id(X))

X = X@Y
print(id(X), end="\n\n")
```

```python
X1 = torch.randint(1, 10, (3, 2, 4))
Y1 = torch.randint(1, 10, (3, 4, 4))

#be careful of the size match
print(id(X1))

X1[ : ] = X1 @ Y1

print(id(X1), end="\n\n")
```
