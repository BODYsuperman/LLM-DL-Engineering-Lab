# Pooling layer

```python
import matplotlib.pyplot as plt
import torch

#read picture
img = plt.imread('../data/duck.jpg')
print("the shape of img is {}".format(img.shape))
```

```python
#transfer data to tensor and change shape to (C, H , W)
input = torch.tensor(img).permute(2, 0, 1).float()
print("the shape of input is {}".format(input.shape))
```

```python
#initialize the convolutional kernel

conv = torch.nn.Conv2d(in_channels=3, out_channels=3, kernel_size=9, stride=3, padding=0, bias=False)

conv_out = conv(input)
print("the output shape is {}".format(conv_out.shape))
```

```python
#initialize the pooling layer

pool = torch.nn.MaxPool2d(kernel_size=6, stride=6, padding=1)
```

```python
#pool process
pool_out = pool(conv_out)
print("the output shape is {}".format(pool_out.shape))
```

```python
#convert output to picture
output1 = torch.clamp(conv_out.int(), 0, 255)
output1 = output1.permute(1, 2, 0).detach().numpy()

output2 = torch.clamp(pool_out.int(), 0, 255)
output2 = output2.permute(1, 2, 0).detach().numpy()

fig, ax = plt.subplots(1, 3, figsize=  (12, 4))

ax[0].imshow(img)
ax[1].imshow(output1)
ax[2].imshow(output2)

plt.axis('off')
plt.show()
```
