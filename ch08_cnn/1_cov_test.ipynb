## Definition and Application Testing of Convolutional Layers

```python

import matplotlib.pyplot as plt
import torch

#read picture
img = plt.imread('../data/duck.jpg')
print("the shape of img is {}".format(img.shape))
```

```python
#transfer data to tensor and change shape to (C, H , W)
input = torch.tensor(img).permute(2, 0, 1).float()
print("the shape of input is {}".format(input.shape))
```

```python
#initialize the convolutional kernel

conv = torch.nn.Conv2d(in_channels=3, out_channels=3, kernel_size=9, stride=3, padding=0, bias=False)

output = conv(input)
print("the output shape is {}".format(output.shape))
```

```python
#convert output to picture
output = torch.clamp(output.int(), 0, 255)
output = output.permute(1, 2, 0).detach().numpy()

fig, ax = plt.subplots(1, 2, figsize=  (12, 4))

ax[0].imshow(img)
ax[1].imshow(output)

plt.axis('off')
plt.show()
```
